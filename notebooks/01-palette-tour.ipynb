# plumage — palette tour

A quick tour of the library: apply the theme, look at the palette, and draw a
couple of charts. Notebooks are for exploration only — reusable code lives in
`src/plumage/`.

```python
import plumage as pl
from plumage.gallery import swatches

pl.use_theme()
swatches();
```

```python
df = pl.datasets.monthly_channels()
pl.line(df.index, {c: df[c] for c in df.columns},
        title='Monthly sessions', ylabel='Thousands');
```

```python
m = pl.datasets.measurements()
pl.correlation(m, title='Correlation');
```

Every chart helper returns a matplotlib `Axes`, so you can keep styling or
save it: `ax = pl.scatter(m['x'], m['y'], groups=m['group']); ax.figure.savefig(...)`.
